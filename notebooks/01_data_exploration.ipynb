## 13. Outlier Analysis

Outlier analysis is used to identify observations that are unusually
high or low compared with the majority of the data.

For this project, potential outliers will be investigated in:

### Users
- `age`

### Flights
- `price`
- `time`
- `distance`

### Hotels
- `days`
- `price`
- `total`

Outliers will be identified but will not automatically be removed.
An extreme value may represent a legitimate travel transaction.

```python
def outlier_summary(df, columns):

    results = []

    for column in columns:

        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outlier_mask = (
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        )

        results.append({
            "Column": column,
            "Q1": Q1,
            "Q3": Q3,
            "IQR": IQR,
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound,
            "Outlier Count": outlier_mask.sum(),
            "Outlier %": round(
                outlier_mask.mean() * 100, 2
            )
        })

    return pd.DataFrame(results)
```

```python
user_outliers = outlier_summary(
    users,
    ["age"]
)

display(user_outliers)
```

```python
flight_outliers = outlier_summary(
    flights,
    ["price", "time", "distance"]
)

display(flight_outliers)
```

```python
hotel_outliers = outlier_summary(
    hotels,
    ["days", "price", "total"]
)

display(hotel_outliers)
```

```python
user_flight_summary = (
    flights
    .groupby("userCode")
    .agg(
        flight_count=("travelCode", "count"),
        avg_flight_price=("price", "mean"),
        avg_distance=("distance", "mean"),
        avg_flight_time=("time", "mean"),
        total_flight_spend=("price", "sum"),
        unique_destinations=("to", "nunique"),
        unique_origins=("from", "nunique")
    )
    .reset_index()
)

display(
    user_flight_summary.head()
)
```

```python
user_hotel_summary = (
    hotels
    .groupby("userCode")
    .agg(
        hotel_booking_count=("travelCode", "count"),
        avg_hotel_price=("price", "mean"),
        avg_hotel_days=("days", "mean"),
        total_hotel_spend=("total", "sum"),
        unique_hotels=("name", "nunique"),
        unique_locations=("place", "nunique")
    )
    .reset_index()
)

display(
    user_hotel_summary.head()
)
```

```python
user_behavior = users.merge(
    user_flight_summary,
    left_on="code",
    right_on="userCode",
    how="left"
)

user_behavior = user_behavior.merge(
    user_hotel_summary,
    left_on="code",
    right_on="userCode",
    how="left",
    suffixes=("_flight", "_hotel")
)

display(
    user_behavior.head()
)
```

```python
hotels["calculated_total"] = (
    hotels["days"] * hotels["price"]
)

hotels["total_difference"] = (
    hotels["total"] -
    hotels["calculated_total"]
)

display(
    hotels["total_difference"].describe()
)
```
