# Travel Data Science & MLOps Capstone
## 02 — Data Preprocessing & Feature Engineering

### Objective

The objective of this notebook is to transform the raw Users,
Flights, and Hotels datasets into clean and machine-learning-ready
datasets.

The preprocessing workflow includes:

- Loading raw datasets
- Data validation
- Data type conversion
- Duplicate and missing-value handling
- Date feature engineering
- Categorical feature preparation
- User behavioral feature engineering
- Dataset integration
- Saving processed datasets

The processed datasets will be used in the subsequent machine
learning notebooks.

```python
import numpy as np
import pandas as pd

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

print("Libraries imported successfully.")
```

```python
users = pd.read_csv("/content/users.csv")
flights = pd.read_csv("/content/flights.csv")
hotels = pd.read_csv("/content/hotels.csv")

print("Datasets loaded successfully.")
```

```python
print("=" * 60)
print("RAW DATASET DIMENSIONS")
print("=" * 60)

print(f"Users   : {users.shape[0]:,} rows × {users.shape[1]} columns")
print(f"Flights : {flights.shape[0]:,} rows × {flights.shape[1]} columns")
print(f"Hotels  : {hotels.shape[0]:,} rows × {hotels.shape[1]} columns")
```

```python
users_clean = users.copy()
flights_clean = flights.copy()
hotels_clean = hotels.copy()
```

```python
print(users_clean.info())
print(flights_clean.info())
print(hotels_clean.info())
```

```python
print(users_clean.isnull().sum())

print(flights_clean.isnull().sum())

print(hotels_clean.isnull().sum())
```

```python
print("Users columns:")
print(users_clean.columns.tolist())

print("\nFlights columns:")
print(flights_clean.columns.tolist())

print("\nHotels columns:")
print(hotels_clean.columns.tolist())
```

```python
# Check duplicates before cleaning
print("Users duplicates:", users_clean.duplicated().sum())
print("Flights duplicates:", flights_clean.duplicated().sum())
print("Hotels duplicates:", hotels_clean.duplicated().sum())

# Remove exact duplicate rows
users_clean = users_clean.drop_duplicates().reset_index(drop=True)
flights_clean = flights_clean.drop_duplicates().reset_index(drop=True)
hotels_clean = hotels_clean.drop_duplicates().reset_index(drop=True)

# Verify
print("\nAfter cleaning")
print("Users duplicates:", users_clean.duplicated().sum())
print("Flights duplicates:", flights_clean.duplicated().sum())
print("Hotels duplicates:", hotels_clean.duplicated().sum())

print("Flights shape:", flights_clean.shape)
```

```python
users_clean = users_clean.drop_duplicates()

flights_clean = flights_clean.drop_duplicates()

hotels_clean = hotels_clean.drop_duplicates()

print(users_clean.shape)
print(flights_clean.shape)
print(hotels_clean.shape)
```

```python
flights_clean["date"] = (
    flights_clean["date"]
    .astype(str)
    .str.strip()
)

hotels_clean["date"] = (
    hotels_clean["date"]
    .astype(str)
    .str.strip()
)
```

```python
flights_clean["date"] = pd.to_datetime(
    flights_clean["date"],
    errors="coerce",
    format="mixed"
)

hotels_clean["date"] = pd.to_datetime(
    hotels_clean["date"],
    errors="coerce",
    format="mixed"
)
```

```python
print(
    "Invalid Flight Dates:",
    flights_clean["date"].isna().sum()
)

print(
    "Invalid Hotel Dates:",
    hotels_clean["date"].isna().sum()
)
```

```python
users_clean["age"] = pd.to_numeric(
    users_clean["age"],
    errors="coerce"
)

for col in ["price", "time", "distance"]:

    flights_clean[col] = pd.to_numeric(
        flights_clean[col],
        errors="coerce"
    )

for col in ["days", "price", "total"]:

    hotels_clean[col] = pd.to_numeric(
        hotels_clean[col],
        errors="coerce"
    )
```

```python

flights_clean["year"] = flights_clean["date"].dt.year

flights_clean["month"] = flights_clean["date"].dt.month

flights_clean["day"] = flights_clean["date"].dt.day

flights_clean["day_of_week"] = flights_clean["date"].dt.dayofweek

flights_clean["is_weekend"] = (
    flights_clean["day_of_week"] >= 5
).astype(int)
```

```python
hotels_clean["year"] = hotels_clean["date"].dt.year

hotels_clean["month"] = hotels_clean["date"].dt.month

hotels_clean["day"] = hotels_clean["date"].dt.day

hotels_clean["day_of_week"] = hotels_clean["date"].dt.dayofweek

hotels_clean["is_weekend"] = (
    hotels_clean["day_of_week"] >= 5
).astype(int)
```

```python
hotels_clean["calculated_total"] = (
    hotels_clean["days"]
    * hotels_clean["price"]
)

hotels_clean["difference"] = (
    hotels_clean["total"]
    - hotels_clean["calculated_total"]
)

display(
    hotels_clean["difference"].describe()
)
```

```python
hotels_clean.drop(
    columns=[
        "calculated_total",
        "difference"
    ],
    inplace=True
)
```

```python
user_flight_features = (
    flights_clean
    .groupby("userCode")
    .agg(
        flight_count=("travelCode","count"),

        average_flight_price=("price","mean"),

        average_flight_distance=("distance","mean"),

        average_flight_time=("time","mean"),

        total_flight_spend=("price","sum"),

        unique_destinations=("to","nunique"),

        unique_origins=("from","nunique"),

        unique_flight_types=("flightType","nunique"),

        unique_agencies=("agency","nunique")
    )
    .reset_index()
)

display(user_flight_features.head())
```

```python
user_hotel_features = (
    hotels_clean
    .groupby("userCode")
    .agg(
        hotel_booking_count=("travelCode","count"),

        average_hotel_price=("price","mean"),

        average_hotel_days=("days","mean"),

        total_hotel_spend=("total","sum"),

        unique_hotels=("name","nunique"),

        unique_hotel_locations=("place","nunique")
    )
    .reset_index()
)

display(user_hotel_features.head())
```

```python
user_features = users_clean.copy()
```

```python
user_features = user_features.merge(
    user_flight_features,
    left_on="code",
    right_on="userCode",
    how="left"
)
```

```python
user_features = user_features.merge(
    user_hotel_features,
    left_on="code",
    right_on="userCode",
    how="left",
    suffixes=("", "_hotel")
)
```

```python
behavior_cols = [
    "flight_count",
    "average_flight_price",
    "average_flight_distance",
    "average_flight_time",
    "total_flight_spend",
    "unique_destinations",
    "unique_origins",
    "unique_flight_types",
    "unique_agencies",
    "hotel_booking_count",
    "average_hotel_price",
    "average_hotel_days",
    "total_hotel_spend",
    "unique_hotels",
    "unique_hotel_locations"
]

for col in behavior_cols:

    user_features[col] = (
        user_features[col]
        .fillna(0)
    )
```

```python
cols_to_drop = [
    col
    for col in user_features.columns
    if col in ["userCode","userCode_hotel"]
]

user_features.drop(
    columns=cols_to_drop,
    inplace=True
)
```

```python
flight_target = "price"
```

```python
print("flights_clean shape:", flights_clean.shape)
print(
    "flights_clean duplicates:",
    flights_clean.duplicated().sum()
)
```

```python
flight_features = [
    "from",
    "to",
    "flightType",
    "time",
    "distance",
    "agency",
    "year",
    "month",
    "day",
    "day_of_week",
    "is_weekend"
]
```

```python
flight_ml_data = flights_clean[
    ["travelCode"] + flight_features + [flight_target]
].copy()

print("Flight ML dataset shape:", flight_ml_data.shape)

print(
    "Exact duplicate rows:",
    flight_ml_data.duplicated().sum()
)
```

```python
processed_dir = Path(
    "/content/processed"
)

processed_dir.mkdir(
    exist_ok=True
)
```

```python
users_clean.to_csv(
    processed_dir / "users_clean.csv",
    index=False
)

flights_clean.to_csv(
    processed_dir / "flights_clean.csv",
    index=False
)

hotels_clean.to_csv(
    processed_dir / "hotels_clean.csv",
    index=False
)

user_features.to_csv(
    processed_dir / "user_features.csv",
    index=False
)

flight_ml_data.to_csv(
    processed_dir / "flight_ml_data.csv",
    index=False
)
```
